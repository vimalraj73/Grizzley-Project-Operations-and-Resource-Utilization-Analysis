## 6. Keys & Relationships

### Objective

Identify the key columns and relationships between the source tables.

The database relationships are retrieved from PostgreSQL's `information_schema` to understand how the tables are connected.

```python
query = """
SELECT
    tc.table_name AS child_table,
    kcu.column_name AS child_column,
    ccu.table_name AS parent_table,
    ccu.column_name AS parent_column
FROM information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kcu
    ON tc.constraint_name = kcu.constraint_name
    AND tc.table_schema = kcu.table_schema
JOIN information_schema.constraint_column_usage AS ccu
    ON ccu.constraint_name = tc.constraint_name
    AND ccu.table_schema = tc.table_schema
WHERE tc.constraint_type = 'FOREIGN KEY'
  AND tc.table_schema = 'public'
ORDER BY tc.table_name;
"""

relationships = pd.read_sql(query, engine)

relationships
```

### Relationship Overview

The main relationships connect:

- Users to teams through `team_id`
- Projects to teams through `team_id`
- Computed projects to teams through `team_id`
- Subscriptions to teams through `team_id`
- Tasks to projects through `project_id`
- Computed tasks to projects through `project_id`
- Declarations to users through `user_id`
- Declarations to tasks through `task_id`

## 8. Date Coverage

### Objective

Determine the date range available in the date and timestamp columns across the source tables.

For each date-related column, the following metrics are calculated:

- Minimum date
- Maximum date
- Number of missing dates

```python
query = """
SELECT
    table_name,
    column_name,
    data_type
FROM information_schema.columns
WHERE table_schema = 'public'
AND data_type IN ('date', 'timestamp', 'timestamp without time zone', 'timestamp with time zone')
ORDER BY table_name, ordinal_position;
"""

date_tables = pd.read_sql(query, engine)

results = []

for _, row in date_tables.iterrows():
    table_name = row["table_name"]
    column_name = row["column_name"]

    query = f"""
    SELECT
        '{table_name}' AS table_name,
        '{column_name}' AS column_name,
        MIN("{column_name}") AS min_date,
        MAX("{column_name}") AS max_date,
        COUNT(*) - COUNT("{column_name}") AS missing_dates
    FROM "{table_name}";
    """

    results.append(pd.read_sql(query, engine))

results_df = pd.concat(results, ignore_index=True)
results_df
```

### Observations

Most date and timestamp columns contain data covering the period from approximately 2017 to 2024.

The `declarations.date` column contains values ranging from `0001-01-01` to `2030-01-01`, which differs considerably from the date ranges observed in the other tables.

This requires further investigation during the data cleaning stage.
