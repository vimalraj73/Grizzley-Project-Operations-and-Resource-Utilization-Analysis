```python
def table_profile(table_name):

    # Row count
    row_count = pd.read_sql(
        f'SELECT COUNT(*) AS row_count FROM "{table_name}"',
        engine
    ).iloc[0, 0]

    # Only load 5 rows for pandas dtype detection
    sample = pd.read_sql(
        f'SELECT * FROM "{table_name}" LIMIT 5',
        engine
    )

    print(f"Table: {table_name}")
    print(f"Shape: ({row_count:,}, {len(sample.columns)})")

    print("\nColumns:")
    for column in sample.columns:
        print(f"  - {column}: {sample[column].dtype}")
```

```python
def uniqueness_profile(table_name, column):
    query = text(f"""
        SELECT
            COUNT(*) AS total_rows,
            COUNT(DISTINCT "{column}") AS unique_values,
            COUNT(*) - COUNT(DISTINCT "{column}") AS duplicate_count
        FROM "{table_name}";
    """)

    with engine.connect() as conn:
        result = conn.execute(query).mappings().one()

    return dict(result)
```

```python
query = """
SELECT
    tc.table_name AS child_table,
    kcu.column_name AS child_column,
    ccu.table_name AS parent_table,
    ccu.column_name AS parent_column
FROM information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kcu
    ON tc.constraint_name = kcu.constraint_name
    AND tc.table_schema = kcu.table_schema
JOIN information_schema.constraint_column_usage AS ccu
    ON ccu.constraint_name = tc.constraint_name
    AND ccu.table_schema = tc.table_schema
WHERE tc.constraint_type = 'FOREIGN KEY'
  AND tc.table_schema = 'public'
ORDER BY tc.table_name;
"""

relationships = pd.read_sql(query, engine)

relationships
```

```python
def missing_profile(table_name):
    
    columns = pd.read_sql(
        f"""
        SELECT column_name
        FROM information_schema.columns
        WHERE table_schema = 'public'
          AND table_name = '{table_name}'
        ORDER BY ordinal_position;
        """,
        engine
    )

    results = []

    for column in columns["column_name"]:
        
        query = f"""
        SELECT
            '{table_name}' AS table_name,
            '{column}' AS column_name,
            COUNT(*) AS total_rows,
            COUNT(*) - COUNT("{column}") AS null_count,
            ROUND(
                (COUNT(*) - COUNT("{column}")) * 100.0 / COUNT(*),
                2
            ) AS null_percentage
        FROM "{table_name}";
        """

        results.append(pd.read_sql(query, engine))

    return pd.concat(results, ignore_index=True)
```

```python
query = """
SELECT
    table_name,
    column_name,
    data_type
FROM information_schema.columns
WHERE table_schema = 'public'
AND data_type IN ('date', 'timestamp', 'timestamp without time zone', 'timestamp with time zone')
ORDER BY table_name, ordinal_position;
"""

date_tables = pd.read_sql(query, engine)

results = []

for _, row in date_tables.iterrows():
    table_name = row["table_name"]
    column_name = row["column_name"]

    query = f"""
    SELECT
        '{table_name}' AS table_name,
        '{column_name}' AS column_name,
        MIN("{column_name}") AS min_date,
        MAX("{column_name}") AS max_date,
        COUNT(*) - COUNT("{column_name}") AS missing_dates
    FROM "{table_name}";
    """

    results.append(pd.read_sql(query, engine))

results_df = pd.concat(results, ignore_index=True)
results_df
```
